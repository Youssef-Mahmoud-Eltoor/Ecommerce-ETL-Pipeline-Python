# Extract

```python
import pandas as pd
from datetime import datetime

```

```python
# Create Log File
def log(message):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    with open("logfile.txt", "a") as file:
        file.write(f"[{timestamp}] {message}\n")
```

```python

def extract_csvFile(file_path):
    try:
        df = pd.read_csv(file_path, encoding="latin1")
        log("Data extracted successfully")
        print(f"Data extracted successfully: {df.shape}")
        return df

    except Exception as e:
        log(f"Error has been detected: {e}")
        print(f"Error has been detected: {e}")
        return None
```

```python
fileName = 'Dataset/data.csv'
extract_csvFile(fileName)
```

# DATA QUALITY CHECK

```python
# Finding the missing and duplicate records
def data_quality_report(df):

    print("Shape:", df.shape)

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())
```

```python
E_CommerceDF = extract_csvFile(fileName)
log("Data Checked Successfully")
data_quality_report(E_CommerceDF)
```

# TRANSFORM

```python
# Convert and Change the date to date time format
try:
    E_CommerceDF["InvoiceDate"] = pd.to_datetime(
        E_CommerceDF["InvoiceDate"],
        errors="coerce"
    )
    log("InvoiceDate Transformed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

```python
# Split InvoiceDate into different format 
try:
    E_CommerceDF["Year"] = E_CommerceDF["InvoiceDate"].dt.year
    E_CommerceDF["Month"] = E_CommerceDF["InvoiceDate"].dt.month
    E_CommerceDF["MonthName"] = E_CommerceDF["InvoiceDate"].dt.month_name()
    E_CommerceDF["Weekday"] = E_CommerceDF["InvoiceDate"].dt.day_name()
    log("InvoiceDate Splited Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

```python
# Calculate Total Amount
try:
    E_CommerceDF['TotalAmount'] = E_CommerceDF['Quantity'] * E_CommerceDF['UnitPrice']
    log("TotalAmount Calculated Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

```python
# Removing the missing values
try:
    E_CommerceDF.dropna()
    log("Missing Values Removed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

# ANALYZE

## Q1.Which countries have the most orders?

```python
orders_by_country = (
    E_CommerceDF.groupby("Country")[
        "InvoiceNo"].nunique().sort_values(ascending=False)
)
try:
    orders_by_country
    log("Most orders by countries Analyzed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

## Q2.Which months have the highest order volume?

```python
Highest_Orders_In_Months= (
    E_CommerceDF.groupby("Month")["InvoiceNo"]
      .nunique()
      .sort_values(ascending=False)
)
try:
    Highest_Orders_In_Months
    log("Highest orders by months Analyzed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

## Q3.Which weekdays are busiest?

```python
Busiest_Days = (
    E_CommerceDF.groupby("Weekday")["InvoiceNo"]
      .nunique()
      .sort_values(ascending=False)
) 

try:
    Busiest_Days
    log("Busiest Days Analyzed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

## Q4.Which products generate the highest sales?

```python
Highest_Product_sales = (
    E_CommerceDF.groupby("Description")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    )

try:
    Highest_Product_sales
    log("Highest Products Sales Analyzed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

## Q5.Which countries generate the highest revenue?

```python
Highest_Country_Revenue = (
    E_CommerceDF.groupby("Country")["TotalAmount"].sum()
    .sort_values(ascending=False)
)
 
try:
    Highest_Country_Revenue
    log("Highest Country Revenue Analyzed Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```

## LOAD

```python
# save the cleaned data

try:
    E_CommerceDF.to_csv(
    "output/cleaned_orders.csv",
    index=False)
    log("Cleaned Data Loaded Successfully ")
except Exception as e:
    log(f"An Error has beed detected {e}")
```
